Exercises: Level 1
Get user input using input(“Enter your age: ”). If user is 18 or older, give feedback: You are old enough to drive. If below 18 give feedback to wait for the missing amount of years. Output:
Enter your age: 30
You are old enough to learn to drive.
Output:
Enter your age: 15
You need 3 more years to learn to drive.

````python

try:
    age = int(input("Enter your Age: "))
    if age <= 0:
        raise ValueError("Age cannot be negative or 0.")
    else:
        print("You are old enough to learn to drive." if age >= 18 else f"You need {18 - age} more years to learn to drive.")
except ValueError as ve:
    print("Invalid input. Please enter numbers only, like 15 or 30.")
````

Exercises: Level 2
Write a code which gives grade to students according to theirs scores:
```sh
90-100, A
80-89, B
70-79, C
60-69, D
0-59, F
```

````python
import math

grade_table = [
    {"min": 90, "max": 100, "grade": "A"},
    {"min": 80, "max": 89,  "grade": "B"},
    {"min": 70, "max": 79,  "grade": "C"},
    {"min": 60, "max": 69,  "grade": "D"},
    {"min": 0,  "max": 59,  "grade": "F"},
]

def get_grade(num, table):
    for row in table:
        if row["min"] <= num <= row["max"]:
            return row["grade"]
    return None

try:
    num = float(input("Enter your marks : "))
    if num <0 or num > 100:
        raise ValueError("Marks must be between 0 and 100.")
    else:
        decimal_part = num - int(num)
        if decimal_part >= 0.5:
            rounded_marks = math.ceil(num)
        else: 
            rounded_marks = math.floor(num)

        print(f"Your grade is: {get_grade(rounded_marks, grade_table)}")
except ValueError as e:
    print(e)
````

````python
person={
    'first_name': 'Asabeneh',
    'last_name': 'Yetayeh',
    'age': 250,
    'country': 'Finland',
    'is_married': True,
    'skills': ['JavaScript', 'React', 'Node', 'MongoDB', 'Python'],
    'address': {
        'street': 'Space street',
        'zipcode': '02210'
    }
}

# 1) Middle skill
skill = person.get('skills', [])
if skill:
    mid = len(skill) // 2
    print(f"The middle skill is: {skill[mid]}")

# 2) Python skill check
if skill:
    print("Has Python skill:", "Python" in skill)

# 3) Developer title
s = set(skill) # Convert list to set to remove duplicates
if s == {"JavaScript", "React"}:
    print("He is a front end developer")
elif {"React", "Node", "MongoDB"}.issubset(s):
    print("He is a fullstack developer")
elif {"Node", "Python", "MongoDB"}.issubset(s):
    print("He is a backend developer")
else:
    print("unknown title")

# 4) Married and lives in Finland
if person.get("is_married") and person.get("country") == "Finland":
    print(f'{person["first_name"]} {person["last_name"]} lives in Finland. He is married.')
````
