```python
#5. Write a function called check-season, it takes a month parameter and returns the season: Autumn, Winter, Spring or Summer.

def check_season(month):
    if not isinstance(month, str) or not month.isalpha():
        return "Invalid input. Please enter a valid month name in full."
    else:
        month = month.lower()
        if month in ['november','december', 'january', 'february']:
            return "Winter"
        elif month in ['march', 'april', 'may', 'june']:
            return "Summer"
        elif month in ['july', 'august', 'september', 'october']:
            return "Autumn"

check_season("March015")
```

```python
# 6. Write a function called calculate_slope which return the slope of a linear equation

def calculate_slope(x1, y1, x2, y2):
    if x2 - x1 == 0:
        return "Slope is undefined (vertical line)"
    return (y2 - y1) / (x2 - x1)

calculate_slope(1, 2, 3, 4)
```

```python
# 7. Quadratic equation is calculated as follows: ax² + bx + c = 0. 
# Write a function which calculates solution set of a quadratic equation, solve_quadratic_eqn.

def solve_quadratic_eqn(a, b, c):
    if a== 0:
        return "Coefficient 'a' cannot be zero for a quadratic equation."
    discriminant = b**2 - 4*a*c
    if discriminant < 0:
        return "No real roots"
    elif discriminant == 0:
        return -b / (2*a)
    else:
        root1 = (-b + discriminant**0.5) / (2*a)
        root2 = (-b - discriminant**0.5) / (2*a)
        return root1, root2

print(solve_quadratic_eqn(2, 2, 0))  # Example usage
```

```python
# 9. Declare a function named reverse_list. It takes an array as a parameter and it returns the reverse of the array (use loops).
# print(reverse_list([1, 2, 3, 4, 5]))
# [5, 4, 3, 2, 1]
# print(reverse_list(["A", "B", "C"])) 
# ["C", "B", "A"]

def reverse_list(arr):
    if not isinstance(arr, list):
        return "Input must be a list."
    reversed_arr = []
    for i in range(len(arr)-1, -1, -1):
        reversed_arr.append(arr[i])
    return reversed_arr

print(reverse_list([1, 2, 3, 4, 5]))  # Example usage
```

```python
# 10. Declare a function named capitalize_list_items. It takes a list as a parameter and it returns a capitalized list of items

def capitalize_list_items(lst):
    capitalized_list = []
    if not isinstance(lst, list):
         return "Provide a list of text items"
    else:
        for items in lst:
            if not items.isalpha():
                return "Provide a list of text items"
                break
            else:
                capitalized_list.append(items.upper())
        return capitalized_list

capitalize_list_items(["ALPHA", "beta", "gammA"])
```

```python
# 12. Declare a function named remove_item. It takes a list and an item parameters. It returns a list with the item removed from it.
# food_stuff = ['Potato', 'Tomato', 'Mango', 'Milk']
# print(remove_item(food_stuff, 'Mango'))  # ['Potato', 'Tomato', 'Milk'];
# numbers = [2, 3, 7, 9]
# print(remove_item(numbers, 3))  # [2, 7, 9]

def remove_item(lst, item):
    if not isinstance(lst, list):
        return "First argument must be a list."
    if item in lst:
        lst.remove(item)
    return lst

food_stuff = ['Potato', 'Tomato', 'Mango', 'Milk']
print(remove_item(food_stuff, 'Mango'))  # ['Potato', 'Tomato', 'Milk']
```

```python
# 13. Declare a function named sum_of_numbers. It takes a number parameter and it adds all the numbers in that range.
# print(sum_of_numbers(5))  # 15
# print(sum_of_numbers(10)) # 55
# print(sum_of_numbers(100)) # 5050

def sum_of_numbers(n):
    if not isinstance(n, int) or n < 0:
        return "Input must be a non-negative integer."
    total = 0
    for i in range(n + 1):
        total += i
    return total

print(sum_of_numbers(5))  # 15
```

```python
# 1. Declare a function named evens_and_odds . It takes a positive integer as parameter and 
# it counts number of evens and odds in the number.
#     print(evens_and_odds(100))
    # The number of odds are 50.
    # The number of evens are 51.

def evens_and_odds(n):
    if not isinstance(n, int) or n< 0:
        return "Input must be a positive integer."
    else:
        evens = 0
        odds = 0
        for i in range(n+1):
            if i % 2 == 0:
                evens +=1
            else:
                odds +=1
    return f"The number of odds are {odds}.\nThe number of evens are {evens}."

print(evens_and_odds(100))
```

```python
# 2. Call your function factorial, it takes a whole number as a parameter and it return a factorial of the number

def factorial(n: int) -> int:
    if not isinstance(n, int) or n < 0:
        raise ValueError("n must be a non-negative integer")

    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

print(factorial(5))  # 120
print(factorial(0))  # 1
```

```python
# Write a functions which checks if all items are unique in the list.
def is_unique(lst):
    if not isinstance(lst, list):
        raise ValueError("Input must be a list.")
    unique_items = []
    for item in lst:
        if item not in unique_items:
            unique_items.append(item)

    return unique_items

print(is_unique([1, 2, 3, 4, 5, 5]))  # Example usage
print(is_unique(["a", "b", "a", "c", "b"]))       # ['a', 'b', 'c']
print(is_unique([1, "a", 1, "b", "a"])) 
print(is_unique([1, [2, 3], 1, [2, 3], [4, 5]]))
```

```python
# Write a function which checks if all the items of the list are of the same data type.
def check_same_data_type(lst):
    if not isinstance(lst, list):
        raise ValueError("Input must be a list.")
    else:
        first_type = type(lst[0])
        for items in lst:
            if type(items) != first_type:
                return "Not all items are of the same data type."
        return "All items are of the same data type."

print(check_same_data_type([1, 2, 3, 4, 5]))  # All items are of the same data type.
print(check_same_data_type([1, "a", 3, 4, 5]))  # Not all items are of the same data type.
```

```python
# Write a function which check if provided variable is a valid python variable
def is_valid_variable(name):
    import keyword
    if not isinstance(name, str):
        return "Input must be a string."
    if not name.isidentifier() or keyword.iskeyword(name):
        return False
    return True

print(is_valid_variable("variable_name"))  # True
print(is_valid_variable("1variable_name"))  # False
print(is_valid_variable("for"))  # False
```

```python
# Go to the data folder and access the countries-data.json file.
# Create a function called the most_spoken_languages in the world. It should return 10 or 20 most spoken languages in the world 
# in descending order.
# Create a function called the most_populated_countries. It should return 10 or 20 most populated countries in descending order.

# show json file first
import json
from pathlib import Path

file_path = Path(r"C:\Users\ChandrG\DE Track\30DaysOfPython\Data\countries_data.json")

try:
    with open(file_path, 'r', encoding='utf-8') as file:
        countries_json = json.load(file)
except FileNotFoundError:
    print(f"File not found: {file_path}")
except json.JSONDecodeError:
    print(f"Error decoding JSON from the file: {file_path}")

print(json.dumps(countries_json[:2], indent=2))  # Print the first country data to verify

#---------------most_spoken_languages------------------------

def most_spoken_languages(data, top_n=10):
    counts = {}

    for country in data:
        languages = country.get("languages", [])

        for language in languages:
            if language not in counts:
                counts[language] = 1
            else:
                counts[language] = counts[language] + 1

    sorted_languages = sorted(counts.items(), key=lambda item: item[1], reverse=True)
    return sorted_languages[:top_n]

display(most_spoken_languages(countries_json, 20))

#---------------most_populated_countries------------------------

def most_populated_countries(data, top_n = 10):
    population_count = {}
    for item in data:
        country = item.get('name', '')
        population = item.get('population', 0)
        population_count[country] = population
    return sorted(population_count.items(), key=lambda x: x[1], reverse=True)[:top_n]

display(most_populated_countries(countries_json, 20))
```
