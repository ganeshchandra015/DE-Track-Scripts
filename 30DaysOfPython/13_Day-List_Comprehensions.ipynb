Filter only negative and zero in the list using list comprehension
numbers = [-4, -3, -2, -1, 0, 2, 4, 6]

```python
numbers = [-4, -3, -2, -1, 0, 2, 4, 6]
new_nums = [x for x in numbers if x <=0]
print(new_nums)  # Output: [-4, -3, -2, -1, 0]
```

Flatten the following list of lists of lists to a one dimensional list :
list_of_lists =[[1, 2, 3], [4, 5, 6], [7, 8, 9]]

output
[1, 2, 3, 4, 5, 6, 7, 8, 9]

```python
list_of_lists =[[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened_list = [x for sublist in list_of_lists for x in sublist]
print(flattened_list)  # Output: [1, 2, 3, 4, 5, 6, 7, 8, 9]
```

Flatten the following list to a new list:
countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]
output:
[['FINLAND','FIN', 'HELSINKI'], ['SWEDEN', 'SWE', 'STOCKHOLM'], ['NORWAY', 'NOR', 'OSLO']]

```python
countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]
output = [ [item.upper(), item[:3].upper(), city.upper()] for sublist in countries for (item, city) in sublist]
output
```

Change the following list of lists to a list of concatenated strings:
countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]
output:
[{'country': 'FINLAND', 'city': 'HELSINKI'},
{'country': 'SWEDEN', 'city': 'STOCKHOLM'},
{'country': 'NORWAY', 'city': 'OSLO'}]

```python
countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]
output = [{'country': country.upper(), 'city': city.upper()} for sublist in countries for (country, city) in sublist]

output
```

Write a lambda function which can solve a slope or y-intercept of linear functions.

```python
line_params = lambda x1, y1, x2, y2: (
    {'slope': None, 'intercept': None} if x1 == x2
    else 
    {'slope': (y2 - y1) / (x2 - x1), 'intercept': y1 - ((y2 - y1) / (x2 - x1)) * x1}
)

print(line_params(2, 3, 5, 11))  # {'slope': 2.6666666666666665, 'intercept': -2.333333333333333}
```
