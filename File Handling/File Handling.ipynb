#### 1. Problem Statement: Write a Python program that accepts a user’s name as input and writes it to a file called user.txt.

#### 9. Problem Statement: Write a Python program that attempts to open a file called missing.txt and gracefully handles the case where the file does not exist using a try-except block.

Given Input: The filename missing.txt is hardcoded. The file does not exist in the current directory.

Expected Output: Error: The file was not found.

```python
# Wrap the open() call inside a try block.
# Catch the FileNotFoundError exception in the except clause and print a user-friendly message.
# You can also catch the broader IOError or OSError to handle other file-related issues like permission errors.
# Optionally, add a finally block to run cleanup code regardless of whether an error occurred.

try:
    with open("missing.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("The file 'missing.txt' was not found. Please check the file name and try again.")
except IOError as e:
    print(f"An I/O error occurred: {e}")
finally:
    print("File handling operation completed.")
```

#### 18. Problem Statement: Write a Python program to read a text file and count how many times each letter of the alphabet appears in it.

Given Input: A file words.txt containing: Python is a powerful programming language

Expected Output: h: 1
e: 1
l: 3
o: 2
w: 1
r: 1
d: 1

```python
freq = {}
with open("data.txt", "r") as f:
    content = f.read().lower()

for char in content:
    if char.isalpha():  # Consider only alphabetic characters
        freq[char] = freq.get(char, 0) + 1 # get() method is used to retrieve the current count of the character from the freq dictionary. If the character is not yet in the dictionary, it returns 0 by default. Then we add 1 to this count to update the frequency of that character.

for letter, count in freq.items():
    print(f"'{letter}': {count}")
```

#### 26. Problem Statement: Write a Python program to rename all .txt files in a directory by adding a specified prefix to each filename.

Given Input: A directory containing notes.txt and data.txt, with the prefix 2024_.

Expected Output: 
- Renamed: notes.txt -> 2024_notes.txt
- Renamed: data.txt -> 2024_data.txt

```python
import os

directory = "." # Current directory
prefix = "2026_"

for file_name in os.listdir(directory):
    if file_name.endswith(".txt"):
        old_path = os.path.join(directory, file_name) # os.path.join() is used to create a full file path by combining the directory and the file name. This ensures that the code works correctly across different operating systems, as it handles the appropriate path separators (e.g., "/" on Unix/Linux and "\" on Windows).
        new_path = os.path.join(directory, prefix + file_name) # Create the new file path with the prefix added to the original file name.
        os.rename(old_path, new_path) # Rename the file from old_path to new_path
        print(f"Renamed '{file_name}' to '{prefix + file_name}'")
```

#### 28. Problem Statement: Write a Python program to read the contents of two separate text files and write their combined content into a single new file.

Given Input: Two files – file1.txt containing Hello from file 1. and file2.txt containing Hello from file 2.

Expected Output: A new file merged.txt containing both lines, confirmed by printing Files merged into merged.txt

```python
input_files = ["file1.txt", "file2.txt"]
output_file = "merged.txt"

with open(output_file, "w") as outfile:
    for name in input_files:
        with open(name, "r") as infile:
            outfile.write(infile.read())  # Read the entire content of the input file and write it to the output file
            outfile.write("\n")  # Add a newline between files for separation

print(f"Files merged into {output_file}")
```

#### 30. Problem Statement: Write a Python program to list all files (not subdirectories) in a given directory and save their names to a text file.

Given Input: A directory my_folder containing notes.txt, data.csv, and report.pdf.

Expected Output: A file file_list.txt containing one filename per line, confirmed by printing File list saved to file_list.txt

```python
import os

directory = "./my_folder" # Specify the directory you want to list
output_file = "file_list.txt"

with open(output_file, "w") as f:
    for file_name in os.listdir(directory):
        full_path = os.path.join(directory, file_name) # Create the full path to the file
        if os.path.isfile(full_path):  # Check if it's a file (not a directory)
            f.write(full_path  + "\n")  # Write the file path to the output file

print(f"File paths listed in {output_file}")
```

#### 31. Problem Statement: Write a Python program to read a binary image file and write its contents to a new file, producing an exact duplicate of the original image.

Given Input: An existing image file photo.jpg on disk.

Expected Output: A new file photo_copy.jpg created on disk, confirmed by printing Image copied to photo_copy.jpg

```python
source = "photo.png"
destination = "photo_copy.jpg"
chunk_size = 4096 # 4096 bytes (4 KB) is a common chunk size that balances memory usage and performance for file copying operations. You can adjust this value based on the expected file sizes and available system resources.

with open(source, "rb") as src, open(destination, "wb") as dst:
    while True: # This loop will continue until we break it when we reach the end of the file.
        chunk = src.read(chunk_size) # Read a chunk of data from the source file. If the end of the file is reached, this will return an empty bytes object (b''), which evaluates to False in a boolean context.
        if not chunk: # If the chunk is empty, it means we've reached the end of the file, so we break the loop.
            break      # Break out of the loop when we've reached the end of the file
        dst.write(chunk) # Write the chunk of data to the destination file

print(f"Image copied to {destination}")
```

#### 32. Problem Statement: Write a Python program to read a text file, extract all unique words, sort them alphabetically, and print the result.

Given Input: A file paragraph.txt containing: the cat sat on the mat the cat

Expected Output: 
- cat
- mat
- on
- sat
- the

```python
import string
with open("paragraph.txt", "r") as f:
    content = f.read().lower()

words = content.translate(str.maketrans("", "", string.punctuation)).split() # Remove punctuation and split into words
unique_sorted = sorted(set(words)) # Get unique words and sort them alphabetically

for word in unique_sorted:
    print(word)
```

#### 34. Problem Statement: Write a Python program to read a large text file and split it into smaller files, each containing at most 10 lines, saving them as part_1.txt, part_2.txt, and so on.

Given Input: A file large_file.txt containing 25 lines of text.

Expected Output: 
- Created: part_1.txt (10 lines)
- Created: part_2.txt (10 lines)
- Created: part_3.txt (5 lines)

```python
chunk_size = 10

with open("large_file.txt", "r") as f:
    lines = f.readlines() # readlines() reads the entire file into memory as a list of lines. This is not memory efficient for large files, as it can consume a lot of memory if the file is very large. For large files, it's better to read and process the file line by line or in smaller chunks to avoid memory issues.

total_lines = len(lines)

for i , start in enumerate(range(0, total_lines, chunk_size)):
    chunk = lines[start:start + chunk_size] # Get a chunk of lines based on the current index and chunk size
    output_filename = f"part_{i}.txt"
    with open(output_filename, "w") as out:
        out.writelines(chunk) # Write the chunk of lines to a new output file

    print(f"Created: {output_filename} ({len(chunk)} lines)")
```
