#### 1. Problem Statement: Convert the following dictionary into JSON format

Given Input: data = {"key1" : "value1", "key2" : "value2"}

Expected Output: data = {"key1" : "value1", "key2" : "value2"}

````python
import json

data = {"key1" : "value1", "key2" : "value2"}

jsondata = json.dumps(data) # Convert Python dictionary to JSON string
print(jsondata)
print(type(jsondata)) # Check the type of jsondata
# Confirm if jsondata is a json format string
try:    
    json.loads(jsondata) # Try to parse the JSON string
    print("jsondata is a valid JSON format string.")
except json.JSONDecodeError: # Catch JSON decoding errors
    print("jsondata is not a valid JSON format string.")
````

#### 2. Problem Statement: Access the value of key2 from the following JSON

#### 3. Problem Statement: PrettyPrint following JSON data with indent level 2 and key-value separators should be (",", " = ")

Given Input: sampleJson = {"key1": "value1", "key2": "value2"}

Expected Output:
{
  "key1" = "value2",
  "key2" = "value2",
  "key3" = "value3"
}

````python
import json
sampleJson = {"key1" : "value2", "key2" : "value2", "key3" : "value3"}
PrettyPrintedJson = json.dumps(sampleJson, indent=2, separators=(",", " = ")) # Convert a json string to a pretty printed json string
print(PrettyPrintedJson)
````

#### 4. Problem Statement: Sort JSON keys in and write them into a file

Given Input: Sort following JSON data alphabetical order of keys
```
sampleJson = {"id" : 1, "name" : "value2", "age" : 29}
```

Expected Output: 
```python
    {
        "age": 29,
        "id": 1,
        "name": "value2"
    }
```

````python
import json
sampleJson = {"id" : 1, "name" : "value2", "age" : 29}
print("Start writing json data into a file")
with open("sampleJson.json", "w") as file: # Open a file in write mode
    json.dump(sampleJson, file, indent= 4 , sort_keys=True) # Write the JSON data to the file where dump() is used to write JSON data to a file object where indent is used to specify the indentation level for pretty-printing and sort_keys is used to sort the keys in the JSON output alphabetically.

print("Done writing json data into a file")
````

#### 5. Problem Statement: Access the nested key ‘salary’ from the following JSON

Given Input: In code cell

Expected Output: 7000

````python
import json

sampleJson = """{ 
   "company":{ 
      "employee":{ 
         "name":"emma",
         "payble":{ 
            "salary":7000,
            "bonus":800
         }
      }
   }
}"""

data = json.loads(sampleJson) # Convert JSON string to Python dictionary
print(data["company"]["employee"]["payble"]["salary"]) # Access nested JSON data using keys
````

#### 6. Problem Statement: Convert the following Vehicle Object into JSON

Given Input: Input in code

Expected Output: 
```
{
    "name": "Toyota Rav4",
    "engine": "2.5L",
    "price": 32000
}
```

````python
import json
from json import JSONEncoder

class Vehicle:
    def __init__(self, name, engine, price):
        self.name = name
        self.engine = engine
        self.price = price

vehicle = Vehicle("Toyota Rav4", "2.5L", 32000)

# Convert it into JSON format

class VehicleEncoder(JSONEncoder):
    def default(self, obj):
        return obj.__dict__ # Return the object's __dict__ attribute, which is a dictionary containing all the attributes of the object

print("Encode the vehicle object into JSON format")
vehicle_json = json.dumps(vehicle, cls=VehicleEncoder, indent=4) # Convert the vehicle object to JSON format using the custom encoder
print(vehicle_json)
````

#### 7. Problem Statement: Convert the following JSON into Vehicle Object

Given Input: { "name": "Toyota Rav4", "engine": "2.5L", "price": 32000 }
    

Expected Output: For example, we should be able to access Vehicle Object using the dot operator like this.
    ```
    vehicleObj.name, vehicleObj.engine, vehicleObj.price
    ```

````python
import json

class Vehicle: # Define a Vehicle class to represent a vehicle object
    def __init__(self, name, engine, price): # Initialize the Vehicle object with name, engine, and price attributes
        self.name = name # Set the name attribute of the Vehicle object
        self.engine = engine # Set the engine attribute of the Vehicle object
        self.price = price # Set the price attribute of the Vehicle object

def vehicle_decoder(obj): # Define a custom decoder function to convert JSON data into a Vehicle object
    return Vehicle(obj["name"], obj["engine"], obj["price"]) # Create a Vehicle object from the JSON data

vehicle_object = json.loads('{ "name": "Toyota Rav4", "engine": "2.5L", "price": 32000 }', object_hook=vehicle_decoder) # Convert JSON string to Vehicle object using the custom decoder
print(vehicle_object.name) # Access the name attribute of the Vehicle object
print(vehicle_object.engine) # Access the engine attribute of the Vehicle object
print(vehicle_object.price) # Access the price attribute of the Vehicle object
````
